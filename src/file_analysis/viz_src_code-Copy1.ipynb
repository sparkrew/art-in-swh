````python
from pathlib import Path
from IPython.display import Markdown, HTML, display
import json
import uuid


def print_code_file(file_path, language="javascript", copy_to_clipboard=True):
    file_path = Path(file_path)
    code = file_path.read_text(encoding="utf-8")

    display(Markdown(f"### `{file_path}`\n\n```{language}\n{code}\n```"))

    if copy_to_clipboard:
        button_id = f"copy-btn-{uuid.uuid4().hex}"
        code_json = json.dumps(code)

        display(HTML(f"""
        <button id="{button_id}">Copy code to clipboard</button>

        <script>
        document.getElementById("{button_id}").onclick = async function() {{
            const code = {code_json};

            try {{
                await navigator.clipboard.writeText(code);
                this.innerText = "Copied!";
            }} catch (err) {{
                const textarea = document.createElement("textarea");
                textarea.value = code;
                document.body.appendChild(textarea);
                textarea.select();
                document.execCommand("copy");
                document.body.removeChild(textarea);
                this.innerText = "Copied!";
            }}
        }};
        </script>
        """))


path = "/home/rpinter/links/projects/def-baudry/shared/data/artworks/src/"
file_path = path + """
ionescugeorge/website/Tutorials/P5JS/p5.js_video/11.2_p5.js_photoBooth/sketch.js
"""

print_code_file(file_path.replace("\n", ""))
````
